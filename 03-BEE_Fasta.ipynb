```python
def fasta_one_line(f):
    with open(f, 'r') as r:
        h = None
        first = True
        for l in r:
            l = l.strip()
            if l.startswith('>'):
                if not first:
                    print()
                print(l)
                h = l
                first = False
            else:
                print(l, end='')
        if not first and not l.startswith('>'):
            print()

fasta_one_line('multiline_input.fasta')
```
