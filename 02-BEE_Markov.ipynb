```python
import numpy as np

seq = "CTTCATGTGAAAGCAGACGTAAGTCA"
nts = ['A', 'C', 'G', 'T']
cnt = np.zeros((4, 4), dtype=int)

for i in range(len(seq) - 1):
    a = nts.index(seq[i])
    b = nts.index(seq[i+1])
    cnt[a, b] += 1

trans = np.zeros((4, 4))
for i in range(4):
    s = cnt[i].sum()
    trans[i] = cnt[i] / s if s > 0 else 0

print("Transition Matrix (Markov Model):")
print("    A       C       G       T")
for i in range(4):
    print(f"{nts[i]} ", end="")
    for j in range(4):
        print(f"{trans[i, j]:.3f}  ", end="")
    print()
```
